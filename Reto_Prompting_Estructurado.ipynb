## ⚙️ Configuración inicial (ejecuta esta celda primero, no la modifiques)

````python
import re
import unicodedata

# Diccionario donde se guarda el progreso durante la sesión
progreso = {
    "reto_1": False,
    "reto_2": False,
    "reto_3": False,
    "reto_4": False,
    "reto_5": False,
}
xp_total = {"valor": 0}

def normalizar(texto):
    """Quita acentos, pasa a minúsculas y elimina espacios sobrantes."""
    texto = str(texto).strip().lower()
    texto = unicodedata.normalize("NFD", texto)
    texto = "".join(c for c in texto if unicodedata.category(c) != "Mn")
    return texto

def marcar_correcto(clave, xp):
    if not progreso[clave]:
        progreso[clave] = True
        xp_total["valor"] += xp
        print(f"✅ ¡Correcto! +{xp} XP  (Total: {xp_total['valor']}/100)")
    else:
        print(f"✅ ¡Correcto! (Ya habías ganado los XP de este reto: {xp_total['valor']}/100)")

def marcar_incorrecto(mensaje):
    print(f"❌ Aún no. {mensaje}")

print("Entorno listo. ¡Comienza el reto! 🚀")
````

## 🧩 Desafío 1 · Identificación de técnica

Un usuario envía el siguiente prompt a un modelo de lenguaje:

> "Traduce el siguiente texto de español a francés: *La tecnología avanza rápidamente.*"

¿Qué técnica de prompting está utilizando?

- `"A"` → Few-shot Prompting
- `"B"` → Zero-shot Prompting
- `"C"` → Chain-of-Thought (Cadena de Pensamiento)
- `"D"` → Fine-tuning

Escribe tu respuesta como texto (por ejemplo `respuesta_1 = "B"`) y ejecuta la celda.

````python
# ESCRIBE TU RESPUESTA AQUÍ (A, B, C o D)
respuesta_1 = ""

# --- Validación (no modificar) ---
if normalizar(respuesta_1) == "b":
    marcar_correcto("reto_1", 20)
    print("Explicación: no se dio ningún ejemplo previo, solo una instrucción directa. Eso es Zero-shot.")
elif respuesta_1 == "":
    print("✏️ Todavía no has escrito una respuesta.")
else:
    marcar_incorrecto("Recuerda: no hay ejemplos de entrada/salida en el prompt, solo una instrucción directa.")
````

## 🧩 Desafío 3 · Análisis de caso

Quieres clasificar tweets por su tono: Positivo, Negativo o Neutro.

```
Tweet: "El servicio fue pésimo y la comida fría."          → Sentimiento: Negativo
Tweet: "Me encantó la atención, súper recomendados."        → Sentimiento: Positivo
Tweet: "Llegué a las 5 pm como estaba pactado."             → Sentimiento: ______
```

¿Cuál sería la respuesta correcta para completar la última línea, y qué técnica se está aplicando?

- `sentimiento_3` → `"Positivo"`, `"Negativo"` o `"Neutro"`
- `tecnica_3` → `"zero-shot"` o `"few-shot"`

````python
# ESCRIBE TUS RESPUESTAS AQUÍ
sentimiento_3 = ""
tecnica_3 = ""

# --- Validación (no modificar) ---
sentimiento_ok = normalizar(sentimiento_3) == "neutro"
tecnica_ok = normalizar(tecnica_3) == "few-shot"

if sentimiento_3 == "" or tecnica_3 == "":
    print("✏️ Todavía no has completado las dos respuestas.")
elif sentimiento_ok and tecnica_ok:
    marcar_correcto("reto_3", 20)
    print("Explicación: los dos ejemplos anteriores muestran entrada→salida, guiando al modelo (Few-shot). El tweet final no es claramente positivo ni negativo (Neutro).")
else:
    faltan = []
    if not sentimiento_ok:
        faltan.append("el sentimiento")
    if not tecnica_ok:
        faltan.append("la técnica")
    marcar_incorrecto(f"Revisa: {' y '.join(faltan)}.")
````

## 🧩 Desafío 4 · Transforma la técnica

Transforma el siguiente prompt de **Zero-shot** a **Few-shot**:

> "Convierte el siguiente nombre completo en formato 'Apellido, Inicial del Nombre': 'Carlos Mendoza'."

Escribe tu versión Few-shot completa dentro de la variable `prompt_4` (puedes usar varias líneas).
Debe incluir **al menos 2 ejemplos** con el formato `entrada -> salida`, y aplicar el mismo patrón al caso de **Carlos Mendoza**.

Ejemplo de estructura esperada (no copies literalmente, la validación es flexible):
```
Ejemplo 1:
"Ana Torres" -> Torres, A.

Ejemplo 2:
"Roberto Gómez" -> Gómez, R.

Tarea:
"Carlos Mendoza" -> Mendoza, C.
```

````python
# ESCRIBE TU RESPUESTA AQUÍ (puede ser un texto de varias líneas con comillas triples)
prompt_4 = """
"""

# --- Validación flexible (no modificar) ---
texto = normalizar(prompt_4)

tiene_dos_ejemplos = len(re.findall(r"ejemplo", texto)) >= 2
tiene_flechas = len(re.findall(r"->|→", texto)) >= 2
tiene_carlos = "carlos mendoza" in texto
tiene_patron = bool(re.search(r"mendoza,\s*c\.?", texto))

puntos = sum([tiene_dos_ejemplos, tiene_flechas, tiene_carlos])

if prompt_4.strip() == "":
    print("✏️ Todavía no has escrito tu prompt Few-shot.")
elif puntos >= 3 and tiene_patron:
    marcar_correcto("reto_4", 20)
    print("Explicación: incluiste ejemplos de entrada→salida y aplicaste el mismo patrón a Carlos Mendoza → Mendoza, C.")
elif puntos >= 3 and not tiene_patron:
    marcar_incorrecto("Tienes la estructura Few-shot, pero falta aplicar bien el patrón a Carlos Mendoza (debería quedar 'Mendoza, C.').")
else:
    marcar_incorrecto("Te falta estructura Few-shot: incluye al menos 2 ejemplos con formato 'entrada -> salida' y luego la tarea con Carlos Mendoza.")
````

## 🧩 Desafío 5 · Verdadero o falso

> "El Few-shot Prompting consume menos tokens (texto de entrada) que el Zero-shot Prompting, por lo tanto es más económico de procesar."

¿Es verdadero o falso? Escribe `"V"` o `"F"`.

````python
# ESCRIBE TU RESPUESTA AQUÍ ("V" o "F")
respuesta_5 = ""

# --- Validación (no modificar) ---
if normalizar(respuesta_5) == "f":
    marcar_correcto("reto_5", 20)
    print("Explicación: al incluir ejemplos, Few-shot usa MÁS tokens de entrada que Zero-shot, no menos.")
elif respuesta_5 == "":
    print("✏️ Todavía no has escrito una respuesta.")
else:
    marcar_incorrecto("Es falso: Few-shot consume MÁS tokens por incluir ejemplos.")
````
